```python
#import libraries
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider 

#define functions U1, U2
# i dont know how to define what Q1, Q2 are since im not given much, but we can figure it out
# ok. step by step.
 # define the linspace
q1 = np.linspace(-5, 5, 100)
q2 = np.linspace(-5, 5, 100)
    # better too much than too little
Q1, Q2 = np.meshgrid(q1, q2)

def plot_function (K1, K2, C):
    plt.style.use('classic')
    q2 = (-3, 0, 2)
    U1 = 0.5 * K1 * q1**2 + 0.5 * K2 * q2[1]**2 + C * q1 * q2[1]**2
    U2 = 0.5 * K1 * q1**2 + 0.5 * K2 * q2[0]**2 + C * q1 * q2[0]**2
    U3 = 0.5 * K1 * q1**2 + 0.5 * K2 * q2[2]**2 + C * q1 * q2[2]**2
    U4 = 0.5 * K1 * Q1**2 + 0.5 * K2 * Q2**2 + C * Q2 * Q2**2
    plt.figure(figsize=(8,8))
    
    plt.subplot(2, 1, 1)
    #q2 = np.linspace(0, 0, 100)
    plt.plot(U1, color='red', label='$U(Q_1, Q_2=0)$')
    plt.plot(U2, color='blue', label='$U(Q_1, Q_2=-3)$')
    plt.plot(U3, color='grey', label='$U(Q_1, Q_2=2)$')
    #plt.plot(q1, U, color='red')
    plt.xlabel('$Q_1$', fontsize=12)
    plt.ylabel('$Q_2$', fontsize=12)
    plt.legend(fontsize=12)
    
    plt.subplot(2, 1, 2)
    plt.contour(Q1, Q2, U4, levels=20)
    plt.xlabel('$U$', fontsize=12)
    plt.ylabel('$Q_1,Q_2$', fontsize=12)
    plt.colorbar()
 
    
    plt.tight_layout()
    plt.show()


plot_function(1, 1, 1)
#ipy widgets
interact(plot_function,
    K1=FloatSlider(min=-10, max=10, step=0.5, value=1, description='$K_1$'),
    K2=FloatSlider(min=-10, max=10, step=0.5, value=1, description='$K_2$'),
    C=FloatSlider(min=-1, max=1, step=0.01, value=0, description='$C$')      
        )
```
